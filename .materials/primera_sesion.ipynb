## Tipos de datos

### Números

Python permite trabajar con distintos tipos de números. Los principales son:
* Enteros (int)
* Flotantes (float)

### Enteros

Los enteros son números sin parte decimal:

```python
1
2
3
0
-1
-100
```

### Flotantes

Los flotantes representan números con parte decimal:


```python
1.1
2.2
3.3
1.0
-5.4
```

## Operaciones con números
Operaciones aritméticas básicas:

"+"  -> suma

"-" -> resta

"*" -> multiplicacion

"**" -> #potencia

"/" -> division

"//" ->division entera

"%" -> modulo

### Suma

```python
1 + 2
# 3
```

### Resta

```python
2 - 1
# 1
```

### Multiplicación

```python
2 * 2
# 4
```

### Potencia

```python
2 ** 2
# 4

2 ** 3
# 8
```

### División

```python
3 / 2
# 1.5
```

### División entera

```python
7//2
```

```python
8 // 2
# 4
```

```python
-7 // 2
# La división entera obtiene el cociente redondeado hacia abajo
```

### Módulo

```python
7 % 2
# 1  residuo

10 % 3
# 1 
```

Útil para determinar si un número es par

```python
10 % 2
# 0

11 % 2
# 1
```

### División entre 0

```python
2 ** -1
# 0.5
```

```python
10 ** -2
# 0.01
```

```python
# Cualquier número elevado a la 0 es 1
0**0
```

### Booleanos

Los booleanos son un tipo de dato lógico que solo puede tomar dos valores: True (verdadero) o False (falso).

```python
True
False

type(True) 
type(False)
```

```python
# Conversioó a entero

int(True) #1
int(False) #0
```

```python

# Operaciones con booleanos

True + True 
True + False 
False + False

True * 5 #5
False * 5 #0
```

```python
#Conversion a booleano

bool(1) #True
bool(10) #True
bool(-5) #True
bool(0) #False

bool("") #False
bool("Hola") #True
bool("False") #True
bool("0") #True
```

```python
# Comparaciones

True == 1 #True
False == 0 #True

type(True) == type(1) #False
```

```python

# Operadores logicos

True and True #True
True and False #False

True or False #True
False or False #False

not True #False
not False #True
```
