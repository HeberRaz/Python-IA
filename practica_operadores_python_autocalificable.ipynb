# 🧪 Práctica autocalificable — Operadores básicos en Python

Este notebook contiene ejercicios sencillos basados en la guía de **Operadores básicos en Python**.

## Cómo usarlo

1. Ejecuta primero la celda **Inicializar evaluador**.
2. En cada ejercicio modifica **solamente** la celda marcada como `TU CÓDIGO`.
3. Ejecuta la celda de tu código.
4. Ejecuta inmediatamente después la celda **Evaluar**.
5. Cuando termines, ejecuta la sección **Resultado final**.

También puedes usar **Run All / Ejecutar todo**. Cada ejecución vuelve a evaluar el código que hayas modificado.

> No necesitas instalar librerías externas.


```python

# Inicializar evaluador
from IPython.display import display, HTML

class AutoGrader:
    def __init__(self):
        self.results = {}

    def check(self, exercise_id, topic, condition, hint=""):
        passed = bool(condition)
        self.results[exercise_id] = (topic, passed)
        if passed:
            display(HTML(
                "<div style='padding:8px 12px;border-radius:8px;"
                "background:#e9f8ee;border:1px solid #8bd4a1;'>"
                "✅ <b>Correcto</b></div>"
            ))
        else:
            display(HTML(
                "<div style='padding:8px 12px;border-radius:8px;"
                "background:#fff1f0;border:1px solid #efa3a0;'>"
                f"❌ <b>Revisa tu respuesta.</b> Pista: {hint}</div>"
            ))

    def summary(self, total_expected):
        passed = sum(1 for _, ok in self.results.values() if ok)
        checked = len(self.results)
        failed = checked - passed
        pct = round((passed / total_expected) * 100, 1) if total_expected else 0

        by_topic = {}
        for _, (topic, ok) in self.results.items():
            by_topic.setdefault(topic, [0, 0])
            by_topic[topic][1] += 1
            if ok:
                by_topic[topic][0] += 1

        topic_rows = ""
        for topic, (ok_count, total_count) in by_topic.items():
            icon = "✅" if ok_count == total_count else "🟡"
            topic_rows += (
                f"<tr><td>{icon} {topic}</td>"
                f"<td style='text-align:center'>{ok_count}/{total_count}</td></tr>"
            )

        display(HTML(f"""
        <div style="font-family:Arial;padding:18px;border-radius:14px;
                    border:1px solid #d0d7de;background:#fafbfc">
            <h2 style="margin-top:0">📊 Resultado final</h2>
            <div style="font-size:32px;font-weight:bold">{passed}/{total_expected}</div>
            <div style="font-size:18px;margin-bottom:12px">{pct}% correcto</div>
            <p>✅ Correctas: <b>{passed}</b> &nbsp;&nbsp;
               ❌ Incorrectas: <b>{failed}</b> &nbsp;&nbsp;
               🧪 Evaluadas: <b>{checked}/{total_expected}</b></p>
            <table style="border-collapse:collapse;width:100%">
                <tr><th style="text-align:left;border-bottom:1px solid #ddd">Tema</th>
                    <th style="border-bottom:1px solid #ddd">Puntaje</th></tr>
                {topic_rows}
            </table>
        </div>
        """))

grader = AutoGrader()
print("✅ Evaluador inicializado. Ya puedes resolver los ejercicios.")
```
