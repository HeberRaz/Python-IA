````python
# ==========================
# ✏️ TU CÓDIGO
# ==========================

for _ in range(20):

    if right_is_clear():
        turn_right()
        move_forward()

    elif front_is_clear():
        move_forward()

    else:
        turn_left()

    if at_goal():
        break
````
