````python

# Inicializar evaluador
from IPython.display import display, HTML

RESULTADOS = {}

def _mostrar_resultado(numero, correcto, pista=""):
    RESULTADOS[numero] = bool(correcto)

    if correcto:
        display(HTML(
            "<div style='padding:8px 12px;border-radius:8px;"
            "background:#e9f8ee;border:1px solid #8bd4a1;'>"
            f"✅ <b>Pregunta {numero}: Correcto</b></div>"
        ))
    else:
        display(HTML(
            "<div style='padding:8px 12px;border-radius:8px;"
            "background:#fff1f0;border:1px solid #efa3a0;'>"
            f"❌ <b>Pregunta {numero}: Revisa tu respuesta.</b> "
            f"Pista: {pista}</div>"
        ))

def check(numero):
    """
    Evalúa una pregunta por número.

    Uso:
        check(1)
        check(2)
        check(3)
    """

    evaluadores = {

        # Tema 1 — Asignación y desempaquetado
        1:  (lambda: edad == 18, "Asigna el número 18 usando `=`."),
        2:  (lambda: nombre == "Ana", "Recuerda que los textos van entre comillas."),
        3:  (lambda: x == 10 and y == 20, "Puedes asignar dos valores en una sola línea."),

        # Tema 2 — Aritmética y división
        4:  (lambda: suma == 12, "Usa el operador `+`."),
        5:  (lambda: division == 4.5, "En Python `/` produce división real."),
        6:  (lambda: division_entera == 4, "Usa `//` para floor division."),

        # Tema 3 — Enteros grandes y overflow
        7:  (lambda: numero_grande == 10 ** 50, "Python permite enteros muy grandes."),
        8:  (lambda: otro_grande == 10 ** 60, "Puedes combinar `**` y `*`."),
        9:  (lambda: es_entero is True, "Usa `isinstance(valor, int)`."),

        # Tema 4 — Cadenas y f-strings
        10: (lambda: saludo == "Hola, mundo", "Usa `+` entre ambas cadenas."),
        11: (lambda: mensaje == "Hola, Luis", 'Una f-string comienza con `f"..."`.'),
        12: (lambda: frase == "Python es divertido", "No olvides los espacios entre palabras."),

        # Tema 5 — Módulo y operadores unarios
        13: (lambda: residuo == 1, "Usa `%`."),
        14: (lambda: residuo_negativo == 3, "En Python el resultado aquí es 3."),
        15: (lambda: negativo == -7, "Anteponle `-` a la variable."),

        # Tema 6 — Asignación compuesta
        16: (lambda: puntos == 8, "Usa `puntos += 3`."),
        17: (lambda: vidas == 3, "Usa `vidas -= 1`."),
        18: (lambda: doble == 12, "Usa `doble *= 2`."),

        # Tema 7 — Comparaciones y booleanos
        19: (lambda: comparacion_1 is True, "Usa `==`."),
        20: (lambda: comparacion_2 is True, "Usa `>`."),
        21: (lambda: comparacion_3 is False, "La respuesta debe provenir de la comparación."),

        # Tema 8 — Condicionales y tuplas
        22: (lambda: resultado_if == "correcto", 'Usa `if color == "azul":`.'),
        23: (lambda: tupla_menor is True, "Python compara primero el primer elemento."),
        24: (lambda: tuplas_iguales is True, "Usa `==` entre las dos tuplas."),

        # Tema 9 — Identidad y None
        25: (lambda: lista_a == [] and lista_b == [] and mismo_valor is True,
             "Crea `[]` dos veces y compara con `==`."),
        26: (lambda: mismo_objeto is False, "Dos listas creadas por separado no son el mismo objeto."),
        27: (lambda: dato is None and sin_dato is True, "La forma recomendada es `dato is None`."),

        # Tema 10 — Operadores lógicos
        28: (lambda: puede_entrar is True, "Ambas condiciones deben ser verdaderas."),
        29: (lambda: puede_editar is True, "Con `or` basta una condición verdadera."),
        30: (lambda: permitido is True, "Niega el valor de `bloqueado`."),

        # Tema 11 — Pertenencia y rangos
        31: (lambda: contiene_tres is True, "Usa `in`."),
        32: (lambda: no_contiene_diez is True, "Usa `not in`."),
        33: (lambda: numeros == [1, 2, 3, 4], "Convierte el `range` a lista con `list(...)`."),

        # Tema 12 — Potencia
        34: (lambda: potencia_1 == 8, "Usa `**`."),
        35: (lambda: potencia_2 == 25, "Elevar al cuadrado significa usar exponente 2."),
        36: (lambda: potencia_3 == 9, "Usa la función `pow`."),
    }

    if numero not in evaluadores:
        raise ValueError(f"No existe la pregunta {numero}.")

    prueba, pista = evaluadores[numero]

    try:
        correcto = prueba()
        _mostrar_resultado(numero, correcto, pista)
    except Exception as error:
        _mostrar_resultado(
            numero,
            False,
            "Tu código produjo un error: " + str(error)
        )

def resultado_final(total=36):
    correctas = sum(1 for value in RESULTADOS.values() if value)
    evaluadas = len(RESULTADOS)
    incorrectas = evaluadas - correctas
    porcentaje = round((correctas / total) * 100, 1)

    display(HTML(f"""
    <div style="font-family:Arial;padding:18px;border-radius:14px;
                border:1px solid #d0d7de;background:#fafbfc">
        <h2 style="margin-top:0">📊 Resultado final</h2>
        <div style="font-size:32px;font-weight:bold">{correctas}/{total}</div>
        <div style="font-size:18px;margin-bottom:12px">{porcentaje}% correcto</div>
        <p>
            ✅ Correctas: <b>{correctas}</b> &nbsp;&nbsp;
            ❌ Incorrectas: <b>{incorrectas}</b> &nbsp;&nbsp;
            🧪 Evaluadas: <b>{evaluadas}/{total}</b>
        </p>
    </div>
    """))

print("✅ Evaluador inicializado.")
print("Después de cada ejercicio usa: check(numero)")
````

## Para el instructor

La lógica de evaluación está centralizada en una única función:

```python
check(numero)
```

Cada número identifica una pregunta y su prueba correspondiente.

Esto evita repetir bloques `try/except`, llamadas largas al grader y lógica de presentación después de cada ejercicio.

Ejemplos:

```python
check(1)
check(2)
check(3)
```

Para añadir una nueva pregunta basta con registrar una entrada adicional dentro de `evaluadores`.


````python
amigos = ["Eric", "Ana", "Luis", "Marta"]
longitud = len(amigos)
print(f"Longitud de la lista de amigos: {longitud}")

for indice in range(0, longitud):
    print(f"Persona {amigos[indice]} tiene {indice * 3} años")
````
